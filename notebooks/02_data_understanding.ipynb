### $Create$ $a$ $clean$ $torque$ $column$

```python
import re

def clean_torque(value):
    if pd.isna(value):
        return None

    value = str(value).strip()

    # Case 1: Torque already in Nm
    nm_match = re.search(r'([\d.]+)\s*[Nn][Mm]', value)

    if nm_match:
        return float(nm_match.group(1))

    # Case 2: Torque in kgm
    kgm_match = re.search(r'([\d.]+)\s*(?:kgm|kg-m)', value, re.IGNORECASE)

    if kgm_match:
        kgm_value = float(kgm_match.group(1))
        return kgm_value * 9.80665

    # Case 3: Format such as 13.5@ 4,800(kgm@ rpm)
    kgm_format = re.search(r'([\d.]+)\s*@.*\(kgm@', value, re.IGNORECASE)

    if kgm_format:
        kgm_value = float(kgm_format.group(1))
        return kgm_value * 9.80665

    return None
```

```python
import re

def clean_torque(value):
    if pd.isna(value):
        return None

    value = str(value).strip()

    # 1. Explicit Nm formats
    nm_match = re.search(r'([\d.]+)\s*[Nn][Mm]', value)
    if nm_match:
        return float(nm_match.group(1))

    # 2. Explicit kgm formats
    kgm_match = re.search(r'([\d.]+)\s*(?:kgm|kg-m)', value, re.IGNORECASE)
    if kgm_match:
        return float(kgm_match.group(1)) * 9.80665

    # 3. Format: 13.5@ 4,800(kgm@ rpm)
    kgm_format = re.search(
        r'([\d.]+)\s*@.*\(kgm@',
        value,
        re.IGNORECASE
    )
    if kgm_format:
        return float(kgm_format.group(1)) * 9.80665

    # 4. Format: 48@ 3,000+/-500(NM@ rpm)
    nm_format = re.search(
        r'([\d.]+)\s*@.*\(NM@',
        value,
        re.IGNORECASE
    )
    if nm_format:
        return float(nm_format.group(1))

    # 5. Generic numeric torque formats:
    # 250@ 1250-5000rpm
    # 510@ 1600-2400
    # 110(11.2)@ 4800
    # 210 / 1900
    generic = re.match(r'\s*([\d.]+)', value)

    if generic:
        return float(generic.group(1))

    return None
```

```python
numeric_cols = [
    "selling_price",
    "km_driven",
    "mileage",
    "engine",
    "max_power",
    "torque_nm",
    "seats",
    "car_age"
]

outlier_summary = []

for col in numeric_cols:

    Q1 = cars_ml[col].quantile(0.25)
    Q3 = cars_ml[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (cars_ml[col] < lower) |
        (cars_ml[col] > upper)
    ).sum()

    outlier_summary.append({
        "column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower,
        "upper_bound": upper,
        "outlier_count": outliers,
        "outlier_percentage": round(
            outliers / len(cars_ml) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

print(outlier_summary)
```

```python
def predict_car_price(
    year,
    km_driven,
    fuel,
    seller_type,
    transmission,
    owner,
    mileage,
    engine,
    max_power,
    seats,
    torque_nm,
    car_age
):
    
    car = pd.DataFrame([{
        "year": year,
        "km_driven": km_driven,
        "fuel": fuel,
        "seller_type": seller_type,
        "transmission": transmission,
        "owner": owner,
        "mileage": mileage,
        "engine": engine,
        "max_power": max_power,
        "seats": seats,
        "torque_nm": torque_nm,
        "car_age": car_age
    }])
    
    prediction = loaded_pipeline.predict(car)[0]
    
    return round(prediction)
```
