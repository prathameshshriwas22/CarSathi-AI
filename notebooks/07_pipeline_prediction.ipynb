```python
import pandas as pd
import numpy as np
from pathlib import Path

PROCESSED_PATH = Path(
    r"C:\Users\lenovo\Document_local\Data Science NareshIT Course"
    r"\Projects For GITHUB\CarSathi-AI\data\processed\cars_ml.csv"
)

cars_ml = pd.read_csv(PROCESSED_PATH)

# Recreate the feature used by the original notebook.
CURRENT_YEAR = 2025
cars_ml["car_age"] = CURRENT_YEAR - cars_ml["year"]

print("Processed ML dataset loaded.")
print("Shape:", cars_ml.shape)
print("Columns:", cars_ml.columns.tolist())

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestRegressor

numerical_features = [
    "km_driven", "mileage", "engine", "max_power",
    "seats", "torque_nm", "car_age"
]
categorical_features = [
    "fuel", "seller_type", "transmission", "owner"
]
features = numerical_features + categorical_features

X = cars_ml[features]
y = cars_ml["selling_price"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ]
)

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

```python
def predict_car_price(
    year,
    km_driven,
    fuel,
    seller_type,
    transmission,
    owner,
    mileage,
    engine,
    max_power,
    seats,
    torque_nm,
    car_age
):
    
    car = pd.DataFrame([{
        "year": year,
        "km_driven": km_driven,
        "fuel": fuel,
        "seller_type": seller_type,
        "transmission": transmission,
        "owner": owner,
        "mileage": mileage,
        "engine": engine,
        "max_power": max_power,
        "seats": seats,
        "torque_nm": torque_nm,
        "car_age": car_age
    }])
    
    prediction = loaded_pipeline.predict(car)[0]
    
    return round(prediction)
```
