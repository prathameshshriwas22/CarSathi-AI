```python
# Create an outlier summary
```

```python
numeric_cols = [
    "selling_price",
    "km_driven",
    "mileage",
    "engine",
    "max_power",
    "torque_nm",
    "seats",
    "car_age"
]

outlier_summary = []

for col in numeric_cols:

    Q1 = cars_ml[col].quantile(0.25)
    Q3 = cars_ml[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = (
        (cars_ml[col] < lower) |
        (cars_ml[col] > upper)
    ).sum()

    outlier_summary.append({
        "column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower,
        "upper_bound": upper,
        "outlier_count": outliers,
        "outlier_percentage": round(
            outliers / len(cars_ml) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

print(outlier_summary)
```

```python
print("IQR Outlier Analysis")
print("=" * 70)

for col in numerical_features:

    Q1 = cars_ml[col].quantile(0.25)
    Q3 = cars_ml[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = cars_ml[
        (cars_ml[col] < lower) |
        (cars_ml[col] > upper)
    ]

    print(f"\n{col}")
    print(f"Q1           : {Q1:.2f}")
    print(f"Q3           : {Q3:.2f}")
    print(f"IQR          : {IQR:.2f}")
    print(f"Lower bound  : {lower:.2f}")
    print(f"Upper bound  : {upper:.2f}")
    print(f"Outlier count: {len(outliers)}")
```

```python
# Create a Compact Audit Table
```

```python
audit = []

for col in numerical_features:

    Q1 = cars_ml[col].quantile(0.25)
    Q3 = cars_ml[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = (
        (cars_ml[col] < lower) |
        (cars_ml[col] > upper)
    ).sum()

    audit.append({
        "Feature": col,
        "Min": cars_ml[col].min(),
        "Max": cars_ml[col].max(),
        "Mean": cars_ml[col].mean(),
        "Median": cars_ml[col].median(),
        "Skewness": cars_ml[col].skew(),
        "Zero_Count": (cars_ml[col] == 0).sum(),
        "Negative_Count": (cars_ml[col] < 0).sum(),
        "IQR_Outliers": outlier_count
    })

outlier_audit = pd.DataFrame(audit)

print(outlier_audit)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
```
